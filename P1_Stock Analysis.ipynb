```python
# Importing required libraries for ETL, analysis, and visualization
import pandas as pd
import numpy as np
import logging
```

```python
# Setting up logging for ETL pipeline tracking

logging.basicConfig(level=logging.INFO)
logging.info("Stock Analysis ETL Pipeline Started")
```

```python
# EXTRACT STEP: Reading dirty CSV stock datasets
aapl_df = pd.read_csv("aapl_stock_prices.csv")
goog_df = pd.read_csv("goog_stock_prices.csv")
msft_df = pd.read_csv("msft_stock_prices.csv")
```

```python
# Adding symbol column to each dataset to match project DDL structure
aapl_df["symbol"] = "AAPL"
goog_df["symbol"] = "GOOG"
msft_df["symbol"] = "MSFT"
```

```python
# Combining all stock datasets into a single dataframe (as per unified MySQL table requirement)
stock_df = pd.concat([aapl_df, goog_df, msft_df], ignore_index=True)
stock_df
```

```python
# Converting date column to datetime for time series forecasting and trend analysis
stock_df["date"] = pd.to_datetime(stock_df["date"])
```

```python
# Checking missing values in the combined dataset (Null handling step in ETL)
stock_df.isnull().sum()
```

```python
# Removing duplicate records as part of data deduplication in ETL transformation stage
stock_df = stock_df.drop_duplicates()
stock_df
```

```python
# Applying Missing Value Treatment Rule: replace close_price with median close_price per stock symbol
stock_df["close_price"] = stock_df.groupby("symbol")["close_price"].transform(
    lambda x: x.fillna(x.median())
)
stock_df
```

```python
# Applying Missing Value Treatment Rule: replace missing volume values with 0
stock_df["volume"] = stock_df["volume"].fillna(0)
stock_df
```

```python
# Applying business rule: replace missing high_price using max(open_price, close_price)
stock_df["high_price"] = stock_df.apply(
    lambda row: max(row["open_price"], row["close_price"])
    if pd.isnull(row["high_price"]) else row["high_price"], axis=1
)
stock_df
```

```python
# Applying business rule: replace missing low_price using min(open_price, close_price)
stock_df["low_price"] = stock_df.apply(
    lambda row: min(row["open_price"], row["close_price"])
    if pd.isnull(row["low_price"]) else row["low_price"], axis=1
)
stock_df
```

```python
# Price Sanity Check: ensuring high_price is always greater than or equal to open and close price
stock_df["high_price"] = stock_df[["high_price", "open_price", "close_price"]].max(axis=1)

# Price Sanity Check: ensuring low_price is always less than or equal to open and close price
stock_df["low_price"] = stock_df[["low_price", "open_price", "close_price"]].min(axis=1)
stock_df
```

```python
# Creating daily_return column for financial analytics and trend prediction
stock_df["daily_return"] = (
    (stock_df["close_price"] - stock_df["open_price"]) / stock_df["open_price"]
) * 100
stock_df
```

```python
# Trend Classification based on daily_return as per business rule table
stock_df["trend"] = stock_df["daily_return"].apply(
    lambda x: "UP" if x > 0 else ("DOWN" if x < 0 else "NO_CHANGE")
)
stock_df
```

```python
# Sorting data by symbol and date for accurate time series analysis and moving averages
stock_df = stock_df.sort_values(["symbol", "date"])
stock_df
```

```python
# LOAD STEP: Importing SQLAlchemy to connect ETL pipeline with MySQL database
from sqlalchemy import create_engine
```

```python
# Creating MySQL engine using SQLAlchemy (Target layer in ETL architecture)
engine = create_engine("mysql+mysqlconnector://root:Saisree31@localhost:3306/stock_db")
engine
```

```python
# Loading transformed clean data into MySQL stock_prices table (as per given DDL)
stock_df.to_sql(name="stock_prices", con=engine, if_exists="replace", index=False)
print("Clean Stock Data Loaded into MySQL Successfully")
```

```python
# VALIDATION STEP: Reading data from MySQL to verify successful ETL load
pd.read_sql("SELECT * FROM stock_prices", con=engine)
```

```python
# Importing matplotlib for dashboard visualization as mentioned in project tools
import matplotlib.pyplot as plt
```

```python
# Plotting daily closing price trends per stock using group-based visualization (matches project requirement)
plt.figure()
for symbol in stock_df["symbol"].unique():
    temp_df = stock_df[stock_df["symbol"] == symbol]
    plt.plot(temp_df["date"], temp_df["close_price"], label=symbol)

plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.title("Daily Closing Price Trends per Stock")
plt.legend()
plt.show()
```

```python
# Identifying highest volatility stock using standard deviation of daily returns (technical analysis requirement)
volatility_df = stock_df.groupby("symbol")["daily_return"].std().reset_index()
volatility_df
```

```python
plt.subplot(3, 4, 4)
volatility = stock_df.groupby('symbol')['daily_return'].std() * 100
volatility.plot(kind='bar', color=['red', 'orange', 'green'], alpha=0.8)
plt.title('Volatility (Std Dev %)')
plt.ylabel('Volatility %')
plt.xticks(rotation=45)
```

```python
# Selecting stock with highest volatility for investor risk analysis
highest_volatility_stock = volatility_df.loc[
    volatility_df["daily_return"].idxmax(), "symbol"
]
highest_volatility_stock
```

```python
# Calculating volume vs price movement correlation for each company (factor driving stock movement)
correlation_df = stock_df.groupby("symbol").apply(
    lambda x: x["volume"].corr(x["daily_return"])
).reset_index(name="volume_return_correlation")
correlation_df
```

```python
# Calculating 7-day and 30-day moving averages for time series forecasting and trend evaluation
stock_df["MA_7"] = stock_df.groupby("symbol")["close_price"].transform(
    lambda x: x.rolling(window=7).mean()
)

stock_df["MA_30"] = stock_df.groupby("symbol")["close_price"].transform(
    lambda x: x.rolling(window=30).mean()
)
stock_df
```

```python
# Simple bar chart of volatility (std dev)
vol_data = stock_df.groupby('symbol')[['MA_7', 'MA_30']].std().reset_index()
x = np.arange(len(vol_data))
width = 0.35
plt.bar(x - width/2, vol_data['MA_7'], width, label='MA7 Volatility')
plt.bar(x + width/2, vol_data['MA_30'], width, label='MA30 Volatility')
plt.xlabel('Stock')
plt.ylabel('Std Dev ($)')
plt.title('MA Volatility Comparison')
plt.xticks(x, vol_data['symbol'])
plt.legend()

plt.tight_layout()
plt.savefig('ma_charts_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Visualizing moving averages to analyze short-term vs long-term stock trends
plt.figure()
for symbol in stock_df["symbol"].unique():
    temp_df = stock_df[stock_df["symbol"] == symbol]
    plt.plot(temp_df["date"], temp_df["MA_7"], label=f"{symbol} MA 7")
    plt.plot(temp_df["date"], temp_df["MA_30"], linestyle="--", label=f"{symbol} MA 30")

plt.xlabel("Date")
plt.ylabel("Price")
plt.title("7-Day and 30-Day Moving Averages per Stock")
plt.legend()
plt.show()
```
